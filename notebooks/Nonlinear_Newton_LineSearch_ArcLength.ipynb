## 1. Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Single variable Newton Raphson

Find the value of $\sqrt{2}$

```python
tol   = 1e-13        
max_iter = 100 
inital_guess = 1
for i in range(max_iter):
    if i == 0:
        x = inital_guess
    K_i = 2*x
    F_i = x**2 - 5
    delta_x = -F_i / K_i
    x += delta_x
    print(f"Iteration {i}: x = {x:.4e}, F_i = {F_i:.4e}, delta_x = {delta_x:.4e}")
    if abs(delta_x) < tol:
        print(f"Converged at iteration {i}: x = {x:.4e}, F_i = {F_i:.4e}, delta_x = {delta_x:.4e}")
        break



```

## 3. Multi-variable Newton Raphson

$$
\mathbf{F}(\mathbf{x}) =
\begin{bmatrix}
x_1x_2 - 3\\
x_1^3x_2^2 - 9
\end{bmatrix}
=
\mathbf{0}.
$$
The solution is 
$$
\mathbf{x} =
\begin{bmatrix}
1\\
3
\end{bmatrix}
$$



```python
tol   = 1e-7        
max_iter = 100 
inital_guess = [2, 2]
for i in range(max_iter):
    if i == 0:
        x = inital_guess
    K_i = np.array([[x[1], x[0]], [3*x[0]**2*x[1]**2, 2*x[0]**3*x[1]]])
    F_i = np.array([x[0]*x[1] - 3, x[0]**3*x[1]**2 - 9])
    delta_x = np.linalg.solve(K_i, -F_i)
    x += delta_x
    print(f"Iteration {i}: x = {x[0]:.4e}, {x[1]:.4e}, F_i = {F_i[0]:.4e}, {F_i[1]:.4e}, delta_x = {delta_x[0]:.4e}, {delta_x[1]:.4e}")
    if np.linalg.norm(delta_x) < tol:
        print(f"Converged at iteration {i}: x = {x[0]:.4e}, {x[1]:.4e}, F_i = {F_i[0]:.4e}, {F_i[1]:.4e}, delta_x = {delta_x[0]:.4e}, {delta_x[1]:.4e}")
        break
```

## 4. Single variable modifed Newton Raphson

Find the value of $\sqrt{2}$, inital guess $x_0=1$

```python
tol   = 1e-13        
max_iter = 100 
inital_guess = 1
for i in range(max_iter):
    if i == 0:
        x = inital_guess
        K_i = 2*x
    F_i = x**2 - 2 
    delta_x = -F_i / K_i
    x += delta_x



    print(f"Iteration {i}: x = {x:.4e}, F_i = {F_i:.4e}, delta_x = {delta_x:.4e}")
    if abs(delta_x) < tol:
        print(f"Converged at iteration {i}: x = {x:.4e}, F_i = {F_i:.4e}, delta_x = {delta_x:.4e}")
        break
```

# 5. Line search 


```python
def F(x):
    return (3.5*np.sin(x) + x*np.cos(x)/3 + 2*x)*np.exp(0.02*x)
def K(x):
    g  = 3.5*np.sin(x) + x*np.cos(x)/3 + 2*x
    dg = 3.5*np.cos(x) + np.cos(x)/3 - x*np.sin(x)/3 + 2
    return (dg + 0.02*g)*np.exp(0.02*x)
x_vals = np.linspace(0,8,100)
y_vals = F(x_vals)
plt.plot(x_vals, y_vals)
plt.xlabel('x')
plt.ylabel('F(x)')
plt.show()
```

```python

tol   = 1e-13    
fext = 1;    
max_iter = 100 
inital_guess = 1.8
for i in range(max_iter):
    if i == 0:
        x = inital_guess
        
    F_i = F(x)
    K_i = K(x)
    delta_x = (fext-F_i) / K_i

    x +=  delta_x
    print(f"Iteration {i}: x = {x:.4e}, K_i = {K_i:.4e}, delta_x = {delta_x:.4e}")
    if abs(delta_x) < tol:
        print(f"Converged at iteration {i}: x = {x:.4e}, F_i = {F_i:.4e}, delta_x = {delta_x:.4e}")
        break
```

```python
tol   = 1e-13    
fext = 1;    
max_iter = 100 
# inital_guess = 1.9
for i in range(max_iter):
    if i == 0:
        x = inital_guess
        
    F_i = F(x)
    K_i = K(x)
    delta_x = (fext-F_i) / K_i
    # x += delta_x
    # Directional residual; x and delta_x remain fixed during search.
    def G(s):
        return delta_x * (fext - F(x + s * delta_x))
    G0 = G(0.0)
    s = 1.0

    if abs(G(s)) > 0.5 * abs(G0):
        lo, hi = 0.0, 1.0

        for j in range(60):
            s = 0.5 * (lo + hi)
            Gs = G(s)

            if abs(Gs) <= 0.5 * abs(G0):
                break

            if np.sign(Gs) == np.sign(G0):
                lo = s
            else:
                hi = s


    x += s * delta_x

    print(f"Iteration {i}: x = {x:.4e}, F_i = {F_i:.4e}, K_i = {K_i:.4e}, line search s = {s:.4e}, delta_x = {delta_x:.4e}")
    if abs(delta_x) < tol:
        print(f"Converged at iteration {i}: x = {x:.4e}, F_i = {F_i:.4e}, delta_x = {delta_x:.4e}")
        break
```
